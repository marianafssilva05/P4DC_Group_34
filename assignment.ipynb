#### 2) Create the **`friend_merger`** function, which takes two Pandas DataFrames and a list/tuple of strings as input and returns the friends DataFrame enriched with the cols_to_keep columns of df for each of the two friends, using the suffixes _friend1 and _friend2 to distinguish them (e.g. Name_friend1 and Name_friend2).

`````python
%%writefile utils/friend_merger.py
import pandas as pd

def friend_merger(
        friends,
        df,
        cols_to_keep=("Name", "Generation", "Legendary", "Type_Full")
):
    cols = list(cols_to_keep)
    
    # Merge attributes for friend1
    merged_f1 = pd.merge(
        friends,
        df[cols],
        left_on="friend1",
        right_index=True,
        how="left"
    )
    
    # Merge attributes for friend2 and apply suffixes
    merged_both = pd.merge(
        merged_f1,
        df[cols],
        left_on="friend2",
        right_index=True,
        how="left",
        suffixes=("_friend1", "_friend2")
    )
    return merged_both
`````

#### 1) Create the `missing_bool` and `missing_arrays_creator` functions in the same file.

#### `missing_bool` takes a Pandas DataFrame and a column name `col` as input and returns a boolean Series that is True for the rows where that column has a missing value and False otherwise.

#### `missing_arrays_creator` takes a Pandas DataFrame, a boolean Series missing (as produced by missing_bool) and a tuple of column names features as input. The function should return a tuple of two NumPy arrays, containing the values of the feature columns for the rows where missing is True and False, respectively. Both arrays must be min-max normalised with the formula (value - min) / (max - min), applied to each feature separately. 

#### **Be careful**: the min and max must be computed on the entire DataFrame (not separately for each group), so that both arrays share the same scale. The order of the returned arrays matters!

`````python
%%writefile utils/missing_arrays_creator.py
import pandas as pd
import numpy as np

def missing_bool(df, col="Legendary"):
    return df[col].isna()

def missing_arrays_creator(
        df, 
        missing, 
        features=("HP", "Attack", "Defense", "SpAtk", "SpDef", "Speed")
        ):

    # looking for where Legendary is missing and not missing rows
    values_missing = df.loc[missing, list(features)].values
    values_known = df.loc[~missing, list(features)].values

    min_value = df[list(features)].min().values
    max_value = df[list(features)].max().values

    #then we normalize using min and max scale
    values_missing = (values_missing - min_value) / (max_value - min_value)
    values_known = (values_known - min_value) / (max_value - min_value)
    
    return values_missing, values_known
`````

#### 2) Create the `dist_matrix` function, which takes two NumPy arrays M1 (shape n × d) and M2 (shape m × d) as input. The function should return a NumPy array of shape n × m in which the entry at position (i, j) is the Euclidean distance between row i of M1 and row j of M2. 

#### **Do not use loops!** The computation must be vectorised. Only NumPy is needed for this exercise.

`````python
%%writefile utils/dist_matrix.py
import numpy as np

def dist_matrix(M1, M2):
    # Your code goes here
    # We use mathematical equality (a-b)**2 = a**2 + b**2 - 2*a*b
    # 
    # Sum of the squares of the elements of the rows
    sum_M1 = (M1**2).sum(axis=1, keepdims=True)
    sum_M2 = (M2**2).sum(axis=1, keepdims=True)
    # In order to get 2*a*b we do the dot product
    two_M1_M2 = 2*np.dot(M1, M2.T)
    # now we do the sum and sqrt
    return np.sqrt(sum_M1 + sum_M2.T - two_M1_M2)
    
`````

#### 3) Create the **`imputer`** function. The function takes as input a Pandas DataFrame df, a distance matrix dists, the boolean Series missing, an integer k and a column name col. For every row where col is missing, the function must find its k closest known rows according to dists and assign the value chosen by majority vote: True if at least half of the neighbours are True, otherwise False. The function should return a copy of the DataFrame with the missing values filled in, without modifying the original df. 

#### Hint: look at `np.argpartition` to find the k smallest distances efficiently.

`````python
%%writefile utils/imputer.py
import numpy as np

def imputer(df, dists, missing, k=3, col="Legendary"):
    df = df.copy()
    
    known_values = df.loc[~missing, col].to_numpy()
    
    # Get indices of the k smallest distances for each missing row
    k_nearest_idx = np.argpartition(dists, k, axis=1)[:, :k]
    nearest_labels = known_values[k_nearest_idx]
    
    # Majority vote: True if at least half of the k neighbours are True
    imputed_values = np.sum(nearest_labels == True, axis=1) >= (k / 2)
    
    df.loc[missing, col] = imputed_values
    return df
`````

#### 4) Create the `calculate_damage` function, which takes two dictionaries, `attacker` and `defender`, and a `type_chart` dictionary as input. The function should return a tuple (damage, multiplier), where damage is an integer with the HP lost by the defender, and multiplier is the type effectiveness of the attack. Follow these rules:

- #### The attacker uses its higher attack stat: if `attack` >= `sp_atk`, use the attacker’s attack against the defender’s defense. Otherwise, use the attacker’s sp_atk against the defender’s sp_def.
- #### The base damage is ((42 * 60 * `atk` / `dfn`) / 50) + 2, where atk and dfn are the two stats selected above.
- #### Get the type multiplier by calling your **`type_multiplier`** function with the attacker’s and defender’s "types".
- #### Multiply the base damage by the type multiplier and by a random factor between 0.85 and 1.0 (hint: use random.uniform).
- #### The final damage must be an integer and must never be lower than 1.

`````python
%%writefile utils/calculate_damage.py
from type_multiplier import type_multiplier
import random

def calculate_damage(attacker, defender, type_chart):
    # Confirming which attack and deffense to use
    if attacker["attack"] >= attacker["sp_atk"]:
        damage = ((42 * 60 * attacker["attack"] / defender["defense"]) / 50) + 2
    else:
        damage = ((42 * 60 * attacker["sp_atk"] / defender["sp_def"]) / 50) + 2

    multiplier = type_multiplier(attacker["types"], defender["types"], type_chart)
    damage *= multiplier * random.uniform(0.85, 1)
    damage = max(int(damage), 1)
    
    return damage, multiplier
`````

- #### Subtract the damage from the defender's "hp", modifying the defender dictionary directly. The HP must never drop below 0!
- #### Inform the user of who attacked whom and by how much damage. Also write whether the attack was very effective (mult > 1), not very effective (1 > mult > 0) or not effective (mult == 0). Don't print about effectiveness if the multiplier is 1.
- #### Then print the defender’s remaining HP on a new line
#### Notice that the calculations on dictionary must be in-place.

`````python
%%writefile utils/take_turn.py
from calculate_damage import calculate_damage

def take_turn(attacker, defender, type_chart):
    damage, mult = calculate_damage(attacker, defender, type_chart)

    defender["hp"] = defender["hp"] - damage

    if defender["hp"] < 0:
        defender["hp"] = 0 

    print(f'{attacker["name"]} attacks {defender["name"]} for {damage} damage')

    if mult > 1:
        print("Super Effective")
    elif 0 < mult < 1:
        print("Not very effective")
    elif mult == 0:
        print("No effect")

    print(f'{defender["name"]}\'s remaining HP {defender["hp"]}')
`````

#### 6) Create the **`battle`** function, which takes a Pandas DataFrame `df`, two strings `name1` and `name2` and a `type_chart` dictionary as input. The function simulates a full battle between the two Pokemon and returns the dictionary of the winner. Follow these rules:
- #### Obtain both Pokemon with your `get_pokemon` function and print the header for the battle announcing the participants.
- #### The Pokemon with the higher "speed" attacks first. If both have the same speed, the first Pokemon (name1) starts.
- #### Use your `take_turn` function so that the two Pokemon alternate attacks until one of them has 0 HP.
- #### If the defender faints, the battle must end immediately: the fainted Pokemon must not attack back.
- #### Return the dictionary of the Pokemon that still has HP left.

`````python
%%writefile utils/battle.py
from get_pokemon import get_pokemon
from take_turn import take_turn

def battle(df, name1, name2, type_chart):
    pokemon_1 = get_pokemon(df, name1)
    pokemon_2 = get_pokemon(df, name2)


    if pokemon_1["speed"] >= pokemon_2["speed"]: #determine who attacks first
        attacker = pokemon_1
        defender = pokemon_2
    elif pokemon_1["speed"] < pokemon_2["speed"]:
        attacker = pokemon_2
        defender = pokemon_1

    while pokemon_1["hp"] > 0 and pokemon_2["hp"] > 0:
        take_turn(attacker, defender, type_chart)

        if defender["hp"] <= 0: # is the defender was hp < 0 it's the end of the battle
            break 

        attacker, defender = defender, attacker # swtich attacker and defender for the next turn

    if pokemon_1["hp"] > 0:
        return pokemon_1
    else:
        return pokemon_2
`````
